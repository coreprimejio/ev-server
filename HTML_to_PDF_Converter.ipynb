```
# This script converts an HTML file to a PDF file using Pyppeteer.
# Pyppeteer is a Python library to control a headless Chromium browser.
# This is a modern and reliable way to create pixel-perfect PDFs from HTML.

import asyncio
import sys
import os
# Add nest_asyncio to handle environments with a running event loop (like Jupyter)
import nest_asyncio
from pyppeteer import launch

async def convert_html_to_pdf_browser(html_file_path, pdf_file_path):
    """
    Converts a given local HTML file to a PDF file using a headless browser.

    Args:
        html_file_path (str): The path to the input HTML file.
        pdf_file_path (str): The path where the output PDF file will be saved.
    """
    # Ensure the input file exists before launching the browser
    if not os.path.exists(html_file_path):
        print(f"❌ Error: The file '{html_file_path}' was not found.")
        sys.exit(1)

    print(f"🚀 Starting conversion of '{html_file_path}' to '{pdf_file_path}'...")
    print("   (This may take a moment, especially on the first run as it might download Chromium)")

    browser = None
    try:
        # Launch a headless Chromium browser instance
        browser = await launch(
            headless=True,
            args=['--no-sandbox', '--disable-setuid-sandbox']
        )
        page = await browser.newPage()

        # Get the absolute path to the HTML file to ensure it's found correctly.
        # The 'file://' prefix is crucial for loading local files.
        absolute_html_path = f'file://{os.path.abspath(html_file_path)}'

        # Navigate to the local HTML file
        await page.goto(absolute_html_path, {'waitUntil': 'networkidle0'})

        # Generate the PDF
        await page.pdf({
            'path': pdf_file_path,
            'format': 'A4',
            'printBackground': True, # Ensures background colors and images are included
            'margin': {
                'top': '20px',
                'bottom': '20px',
                'left': '20px',
                'right': '20px'
            }
        })

        print(f"✅ Success! PDF saved as '{pdf_file_path}'")

    except Exception as e:
        print(f"❌ An unexpected error occurred: {e}")
        print("   Please ensure you have run 'pip install pyppeteer'.")
        print("   If you have network issues, the automatic Chromium download may have failed.")
        sys.exit(1)
    finally:
        # Ensure the browser is closed even if an error occurs
        if browser:
            await browser.close()

if __name__ == "__main__":
    # Apply nest_asyncio to allow asyncio.run() to be called from a running event loop.
    nest_asyncio.apply()

    # Define the input and output filenames
    input_html = "/Users/ajaygupta/Downloads/online_viewer_net.html"
    output_pdf = "/Users/ajaygupta/Downloads/ratio_and_proportion_guide.pdf"


    # Run the asynchronous conversion function
    # asyncio.run() is the standard way to run an async main function.
    asyncio.run(convert_html_to_pdf_browser(input_html, output_pdf))
```
