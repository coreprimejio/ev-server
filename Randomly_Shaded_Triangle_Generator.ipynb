```
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon
import numpy as np
import random
import math

def create_triangular_division(total_parts, shaded_parts, output_filename="triangle_division.jpg"):
    """
    Generates an image of a large equilateral triangle divided into smaller,
    equal equilateral triangles, with a random subset being shaded.

    Args:
        total_parts (int): The total number of small triangles. Must be a perfect square.
        shaded_parts (int): The number of small triangles to shade randomly.
        output_filename (str): The name for the output image file.
    """
    # 1. Validate that total_parts is a perfect square
    divisions = int(math.sqrt(total_parts))
    if divisions * divisions != total_parts:
        print(f"Error: total_parts must be a perfect square (e.g., 4, 9, 16, 25...). Got {total_parts}.")
        return

    if shaded_parts > total_parts:
        print(f"Error: shaded_parts ({shaded_parts}) cannot be greater than total_parts ({total_parts}).")
        return

    # 2. Define geometry
    side_length = 1.0
    height = side_length * math.sqrt(3) / 2.0
    all_triangles = []

    # 3. Generate a perfect grid of all vertex points
    points = {}
    for row in range(divisions + 1):
        for col in range(row + 1):
            x = (col * side_length) - (row * side_length / 2.0)
            y = -row * height
            points[(row, col)] = (x, y)

    # 4. Construct small triangles by connecting the grid points
    for row in range(divisions):
        for col in range(row + 1):
            # Add the downward-pointing triangle
            p1 = points[(row, col)]
            p2 = points[(row + 1, col)]
            p3 = points[(row + 1, col + 1)]
            all_triangles.append([p1, p3, p2])

            # Add the upward-pointing triangle (if not on the right edge)
            if col < row:
                p1_up = points[(row, col)]
                p2_up = points[(row, col + 1)]
                p3_up = points[(row + 1, col + 1)]
                all_triangles.append([p1_up, p2_up, p3_up])

    # 5. Randomly select triangles to be shaded
    indices = list(range(total_parts))
    shaded_indices = random.sample(indices, shaded_parts)

    # 6. Create the plot and add the small triangles
    fig, ax = plt.subplots(figsize=(10, 10))

    for i, vertices in enumerate(all_triangles):
        # Unshaded parts are grey, shaded parts are blue
        color = 'royalblue' if i in shaded_indices else 'lightgrey'
        triangle = Polygon(vertices, facecolor=color, edgecolor='black', linewidth=1.5)
        ax.add_patch(triangle)

    # 7. Finalize and save the plot
    ax.set_aspect('equal', 'box')
    ax.axis('off') # Hide the x and y axes
    plt.title(f'Triangle with {shaded_parts} of {total_parts} Parts Shaded', fontsize=18)

    # Adjust plot limits to fit the triangle snugly
    main_triangle_base_width = divisions * side_length
    main_triangle_height = divisions * height
    plt.xlim(-main_triangle_base_width / 2 - 0.1, main_triangle_base_width / 2 + 0.1)
    plt.ylim(-main_triangle_height - 0.1, 0.1)

    plt.savefig(output_filename, format='jpg', bbox_inches='tight', pad_inches=0.1)
    plt.close()
    print(f"✅ Chart saved as '{output_filename}'")


# --- Example of how to use the function ---
if __name__ == '__main__':
    # You can change these numbers to whatever you need
    TOTAL_PARTS_IN_GRID = 36  # Must be a perfect square (6*6)
    SHADED_PARTS_COUNT = 11

    create_triangular_division(total_parts=TOTAL_PARTS_IN_GRID,
                               shaded_parts=SHADED_PARTS_COUNT,
                               output_filename="my_triangle_fraction.jpg")
```
